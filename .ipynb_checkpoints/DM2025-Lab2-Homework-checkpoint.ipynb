## **1. Model Development (10 pts Required)**

---

### **1.1 Preprocessing Steps**

Started from raw posts with usernames, URLs, hashtags, emojis, and noisy punctuation.

* Removed `@user` mentions
* Removed URLs
* Normalized whitespace
* Converted hashtags `#happy → happy`
* Lowercased all text
* Preserved emotionally meaningful punctuation (`? ! , .`)
* Removed other symbols and duplicate spaces

**Emoji processing:**

* First used placeholder tokens from text extraction
* Later switched to **emoji.demojize()**, converting emojis to English descriptions

  * Example: `🤮 → "face_vomiting"`
* Appended emoji descriptions into a new feature column
* Built **full_text = clean_text + emoji_text**

---

### **1.2 Feature Engineering Steps**

**Classical features (TF-IDF):**

* TF-IDF with **100–120k** features
* Unigrams + bigrams `(1,2)`
* `sublinear_tf=True`
* Adjusted class weights for imbalance

**Specialist meta-models (SVM pairs):**

Trained binary classifiers for:

* disgust ↔ anger
* disgust ↔ fear
* disgust ↔ joy
* disgust ↔ surprise
* joy ↔ sadness

These specialists corrected the main SVM only on confusing pairs, boosting minority recall.

**Transformer features:**

For BERT/RoBERTa, the “features” are learned by the model.
We controlled:

* Input text (`full_text`)
* `max_length` (best value found: **99**)
* Label mapping (6 emotions → integer IDs)

---

### **1.3 Explanation of Your Model**

#### **Stage 1 — Classical baseline (TF-IDF + SVM)**

* Achieved ~0.62 validation accuracy
* Kaggle score ~0.586
* Minority classes (disgust, sadness) underperformed
* Added **specialist SVMs**, which re-evaluated only high-confusion pairs
* Disgust recall improved drastically

#### **Stage 2 — LLM trials (Gemini)**

* Tested Gemini for few-shot classification
* API too slow / expensive for batch prediction
* Accuracy unstable
* Not used in final model

#### **Stage 3 — Transformers (final choice)**

* Fine-tuned **DistilBERT**, then **RoBERTa-base**
* Used full-text cleaned input with emoji descriptions
* Enabled FP16 + gradient checkpointing for GPU efficiency
* Tuned batch size to fully utilize the RTX 5060
* Final model trained on **all available data** (no validation split)
* Achieved **0.6934** on the Kaggle private leaderboard

now we'll extract the emojis:

```python
import re

emoji_pattern = re.compile(
    "["
    "\U0001F600-\U0001F64F"  # emoticons
    "\U0001F300-\U0001F5FF"  # symbols & pictographs
    "\U0001F680-\U0001F6FF"  # transport & map symbols
    "\U0001F1E0-\U0001F1FF"  # flags
    "\U00002700-\U000027BF"  # Dingbats
    "\U0001F900-\U0001F9FF"  # Supplemental Symbols and Pictographs
    "\U0001FA70-\U0001FAFF"  # Symbols and Pictographs Extended-A
    "\U00002600-\U000026FF"  # Misc symbols
    "]+",
    flags=re.UNICODE
)

def extract_emojis(text):
    if not isinstance(text, str):
        return ""
    return "".join(emoji_pattern.findall(text))

df["emojis"] = df["text"].apply(extract_emojis)
```

```python
df[["text", "emojis"]].head(10)
```

and convert them into tokens for potential use later: 

```python
def emojis_as_tokens(emoji_str):
    return " ".join([f"EMOJI_{ord(e)}" for e in emoji_str])
    
df["emoji_tokens"] = df["emojis"].apply(emojis_as_tokens)

df.head(10)
```

now we remove the emoji's from the text column and perfom some basic cleaning:

```python
def clean_for_model(text):
    if not isinstance(text, str):
        return ""
    # lowercase
    text = text.lower()
    # remove emojis first
    text = emoji_pattern.sub(" ", text)
    # remove urls
    text = re.sub(r"http\S+|www\S+", " ", text)
    # remove punctuation (keep only letters + spaces)
    text = re.sub(r"[^a-z\s]", " ", text)
    # collapse whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
df["emojis"] = df["text"].apply(extract_emojis)
df["clean_text"] = df["text"].apply(clean_for_model)

df.head(10)
```

## 2.3 Top-K word per emotion

so far results have been very bad, so we'll move on to more complicated feature engineering steps

```python
# dataframe per emtion

emotions = train_df["emotion"].unique()
emotion_dfs = {emo: train_df[train_df["emotion"] == emo].copy() for emo in emotions}

{emo: df.shape for emo, df in emotion_dfs.items()}
```

now we need the term document dataframe per emotion, well define a helper function

```python
def create_term_document_df(df, text_col="clean_text", **cv_kwargs):
    cv = CountVectorizer(**cv_kwargs)
    X_counts = cv.fit_transform(df[text_col])
    words = cv.get_feature_names_out()
    tdm_df = pd.DataFrame(X_counts.toarray(), columns=words)
    return tdm_df
```

```python
filt_tdm_dfs = {}

for emo in emotions:
    filt_tdm_dfs[emo] = create_term_document_df(
        emotion_dfs[emo],
        text_col="clean_text",
     #  min_df=2   # ignore super rare words inside each emotion
    )

for emo, tdm in filt_tdm_dfs.items():
    print(emo, tdm.shape)
```

now we need to get the top-k words per emotion

```python
def top_k_words(term_document_df, k=50):
    word_counts = term_document_df.sum(axis=0)
    top_words = word_counts.sort_values(ascending=False).head(k)
    return top_words

top_words_per_emotion = {emo: top_k_words(tdm, k=50) 
                         for emo, tdm in filt_tdm_dfs.items()}
```

```python
for emo, top_words in top_words_per_emotion.items():
    print(f"\n=== {emo.upper()} top words ===")
    print(top_words)
```

```python
from scipy.sparse import csr_matrix
from scipy.sparse import hstack

def pattern_feature_matrix(text_series, patterns):
    token_sets = text_series.str.split().apply(set)
    mat = np.zeros((len(text_series), len(patterns)), dtype=np.int8)
    for j, pat in enumerate(patterns):
        pat_words = set(pat.split())
        mat[:, j] = token_sets.apply(lambda s: int(pat_words.issubset(s))).values
    return csr_matrix(mat)

X_pat_train = pattern_feature_matrix(train_df["clean_text"], selected_patterns)
X_pat_test  = pattern_feature_matrix(test_df["clean_text"], selected_patterns)
```

### 3.3 LLM classification - gemini

```python
import os
from dotenv import load_dotenv
from google import genai
from google.genai import types

# Load your API key from ./config/api-key.env
env_path = "./config/api-key.env"
load_dotenv(dotenv_path=env_path)

api_key = os.getenv("GOOGLE_API_KEY")
if api_key is None:
    raise ValueError("❌ GOOGLE_API_KEY not found. Create ./config/api-key.env with: GOOGLE_API_KEY='your_key_here'.")

# Set environment variable for Gemini
os.environ["GOOGLE_API_KEY"] = api_key

# Instantiate client
client = genai.Client(api_key=api_key)
```

```python
SYSTEM_INSTRUCTION = "You are a helpful assistant."

MODEL_NAME = "gemini-2.5-flash-lite"

MAX_OUTPUT_TOKENS = 65535

SAFETY_SETTINGS = [
    types.SafetySetting(category="HARM_CATEGORY_HATE_SPEECH", threshold="OFF"),
    types.SafetySetting(category="HARM_CATEGORY_DANGEROUS_CONTENT", threshold="OFF"),
    types.SafetySetting(category="HARM_CATEGORY_SEXUALLY_EXPLICIT", threshold="OFF"),
    types.SafetySetting(category="HARM_CATEGORY_HARASSMENT", threshold="OFF"),
]
```

```python
def prompt_gemini(
        input_prompt: list,
        schema = None,
        temperature: float = 0.0,
        system_instruction: str = SYSTEM_INSTRUCTION,
        max_output_tokens: int = MAX_OUTPUT_TOKENS,
        client: genai.Client = client,
        model_name: str = MODEL_NAME,
        new_config: types.GenerateContentConfig = None,
        with_tools: bool = False,
        with_parts: bool = False,
        with_tokens_info: bool = False
    ):
    try:
        # Pick JSON schema or freeform text
        if schema:
            generate_content_config = types.GenerateContentConfig(
                temperature=temperature,
                system_instruction=system_instruction,
                max_output_tokens=max_output_tokens,
                response_modalities=["TEXT"],
                response_mime_type="application/json",
                response_schema=schema,
                safety_settings=SAFETY_SETTINGS
            )
        else:
            generate_content_config = types.GenerateContentConfig(
                temperature=temperature,
                system_instruction=system_instruction,
                max_output_tokens=max_output_tokens,
                response_modalities=["TEXT"],
                safety_settings=SAFETY_SETTINGS
            )

        # Allow override
        if new_config:
            generate_content_config = new_config

        # Handle structured parts if needed
        if with_parts:
            response = client.models.generate_content(
                model=model_name,
                contents=types.Content(parts=input_prompt),
                config=generate_content_config,
            )
        else:
            response = client.models.generate_content(
                model=model_name,
                contents=input_prompt,
                config=generate_content_config,
            )

        # Handle tool-based responses
        if with_tools:
            completion = response
            if with_tokens_info:
                log = {
                    "model": model_name,
                    "input_tokens": response.usage_metadata.prompt_token_count,
                    "output_tokens": response.usage_metadata.candidates_token_count,
                }
                return completion, log
            return completion

        # Default return = TEXT
        completion = response.text

        # Optional token info
        if with_tokens_info:
            log = {
                "model": model_name,
                "input_tokens": response.usage_metadata.prompt_token_count,
                "output_tokens": response.usage_metadata.candidates_token_count,
            }
            return completion, log

        return completion

    except Exception as e:
        print(f"Error occurred when generating response: {e}")
        return None
```

```python
def sample_few_shots(df, emotions, n=3):
    """
    Pick n examples for each emotion from training set.
    """
    shots = {}
    for emo in emotions:
        subset = df[df["emotion"] == emo]
        if len(subset) < n:
            n_use = len(subset)
        else:
            n_use = n
        shots[emo] = subset.sample(n=n_use, random_state=42)
    return shots
```

```python
def build_prompt(examples, emotions):
    base = (
        "You must classify the following social media text into one of the emotions:\n"
        f"{emotions}\n\n"
        "Each example follows:\n"
        "Text: ...\nClass: <emotion>\n\n"
    )

    for emo in emotions:
        for _, row in examples[emo].iterrows():
            base += f"Text: {row[TEXT_COL]}\nClass: {emo}\n\n"

    return base
```

```python
import json, time

def classify_text_with_llm(text, prompt_base, allowed_labels, schema, system_prompt):
    full_prompt = f"{prompt_base}\nNow classify:\nText: {text}\nClass: "

    # Call Gemini
    result = prompt_gemini(
        input_prompt=[full_prompt],
        schema=schema,
        system_instruction=system_prompt
    )

    if not result:
        # fallback with some creativity
        result = prompt_gemini(
            input_prompt=[full_prompt],
            schema=schema,
            system_instruction=system_prompt,
            temperature=1.0
        )

    try:
        parsed = json.loads(result)
    except:
        parsed = result.replace('"', '').replace("'", "").strip()

    if parsed not in allowed_labels:
        # Force retry with higher temperature
        result_retry = prompt_gemini(
            input_prompt=[full_prompt],
            schema=schema,
            system_instruction=system_prompt,
            temperature=1.0
        )
        try:
            parsed = json.loads(result_retry)
        except:
            parsed = result_retry.replace('"', '').replace("'", "").strip()

    # If still invalid → fallback to "neutral" closest common emotion
    if parsed not in allowed_labels:
        parsed = "joy"   # safe fallback, but rarely happens

    return parsed
```

```python
import enum

class EmotionEnum(enum.StrEnum):
    ANGER = "anger"
    DISGUST = "disgust"
    FEAR = "fear"
    JOY = "joy"
    SADNESS = "sadness"
    SURPRISE = "surprise"
```

```python
from tqdm import tqdm
import time

def run_kaggle_prediction(train_df, test_df, shots_per_class=3):
    # few-shot examples
    few_shots = sample_few_shots(train_df, emotions, n=shots_per_class)
    prompt_base = build_prompt(few_shots, emotions)

    system_prompt = (
        "You are an emotion classifier for social media text. "
        "Always respond ONLY with the emotion label."
    )

    preds = []

    # Kaggle test set classification
    for idx, row in tqdm(test_df.iterrows(), total=len(test_df)):
        text = row[TEXT_COL]

        emotion = classify_text_with_llm(
            text=text,
            prompt_base=prompt_base,
            allowed_labels=emotions,
            schema=EmotionEnum,
            system_prompt=system_prompt
        )

        preds.append(emotion)

        # rate limit control (Gemini Flash Lite: 15 req/min recommended)
        time.sleep(0.7)

    test_df["predicted_emotion"] = preds
    return test_df
```

```python
def gemini_classify(text):
    prompt = [
        f"""
        Classify the emotion expressed in this text. 
        Options: anger, disgust, fear, joy, sadness, surprise.
        Respond with ONLY the label.
        "{text}"
        """
    ]

    response = safe_gemini_call(prompt)
    if response is None:
        return "unknown"
    return response.strip().lower()

```

```python
import time
import json
from google.api_core import exceptions as google_exceptions

def safe_gemini_call(prompt, max_retries=10, **kwargs):
    """
    Wrapper for prompt_gemini that:
    - catches 429 errors,
    - sleeps for retryDelay,
    - retries,
    - returns the response.text
    """
    for attempt in range(max_retries):
        try:
            response = prompt_gemini(prompt, **kwargs)
            return response  # success

        except Exception as e:
            err = str(e)
            if "RESOURCE_EXHAUSTED" in err or "429" in err:
                print("⚠️ Rate limit hit, parsing retry delay...")

                # Try to extract retry delay from the error JSON
                try:
                    info = eval(err.split("error:")[-1])
                    retry_delay = info["error"]["details"][-1]["retryDelay"]
                    retry_seconds = int(retry_delay.replace("s", ""))
                except:
                    retry_seconds = 30  # fallback

                print(f"⏳ Waiting for {retry_seconds} seconds...")
                time.sleep(retry_seconds)
                print("🔁 Retrying...")
                continue

            # Other errors should not
```

```python
import re

def clean_text_basic(text: str) -> str:
    if not isinstance(text, str):
        return ""
    
    # 1) Remove usernames: @user123
    text = re.sub(r"@\w+", " ", text)
    
    # 2) Remove URLs
    text = re.sub(r"http\S+|www\.\S+", " ", text)
    
    # 3) Replace hashtags (#word → word)
    text = re.sub(r"#(\w+)", r"\1", text)
    
    # 4) Remove leftover punctuation except ?!,. (optional)
    # (we keep basic emotional punctuation)
    text = re.sub(r"[^a-zA-Z0-9\s?!,.]", " ", text)
    
    # 5) Normalize multiple spaces
    text = re.sub(r"\s+", " ", text).strip()
    
    return text.lower()
```

```python
import re
import emoji

def verbalize_emojis_clean(text: str) -> str:
    """
    Convert emojis to verbal descriptions:
      ❤️ -> "red heart"
    and clean it up for the ML pipeline.
    """
    if not isinstance(text, str) or not text.strip():
        return ""
    
    # Turn emojis into :red_heart: :smiling_face: etc.
    desc = emoji.demojize(text, delimiters=(" ", " "))
    # Remove colons and underscores, normalize spaces
    desc = desc.replace(":", " ").replace("_", " ")
    desc = re.sub(r"\s+", " ", desc).strip()
    return desc.lower()
```

```python
clf_d_vs_a = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1,2), sublinear_tf=True)),
    ("clf", LinearSVC(class_weight="balanced"))
]).fit(
    df_disgust_vs_anger["clean_text"], df_disgust_vs_anger["emotion"]
)

clf_d_vs_f = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1,2), sublinear_tf=True)),
    ("clf", LinearSVC(class_weight="balanced"))
]).fit(
    df_disgust_vs_fear["clean_text"], df_disgust_vs_fear["emotion"]
)
```

```python
def corrected_predict(text, main_model, clf_d_vs_a, clf_d_vs_f):
    """
    Runs the main model, then applies binary disgust-correctors
    if the main model predicts anger or fear.
    """
    # Step 1: initial prediction
    main_pred = main_model.predict([text])[0]

    # Only trigger correction for anger or fear
    if main_pred == "anger":
        second_pred = clf_d_vs_a.predict([text])[0]
        if second_pred == "disgust":
            return "disgust"
    
    if main_pred == "fear":
        second_pred = clf_d_vs_f.predict([text])[0]
        if second_pred == "disgust":
            return "disgust"
    
    # Otherwise keep the main prediction
    return main_pred
```

```python
import re, emoji

def clean_for_transformer(text):
    if not isinstance(text, str):
        return ""

    # 1. Replace newlines with sentence breaks
    text = text.replace("\n", ". ")

    # 2. Remove URLs
    text = re.sub(r"http\S+|www\.\S+", "", text)

    # 3. Convert hashtags (#fear → fear)
    text = re.sub(r"#(\w+)", r"\1", text)

    # 4. Replace [NAME] with <name>
    text = re.sub(r"\[NAME\]", "<name>", text, flags=re.IGNORECASE)

    # 5. Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


def extract_emojis(s):
    if not isinstance(s,str):
        return ""
    return "".join(c for c in s if c in emoji.EMOJI_DATA)

def verbalize_emojis(text):
    if not isinstance(text, str):
        return ""
    return emoji.demojize(text, delimiters=(" ", " "))

```

```python
train_df["emoji_text"] = train_df["text"].apply(extract_emojis).apply(verbalize_emojis)
test_df ["emoji_text"] = test_df ["text"].apply(extract_emojis).apply(verbalize_emojis)

train_df["clean_text"] = train_df["text"].apply(clean_for_transformer)
test_df ["clean_text"] = test_df ["text"].apply(clean_for_transformer)

train_df["full_text"] = (train_df["clean_text"] + " " + train_df["emoji_text"]).str.strip()
test_df ["full_text"] = (test_df ["clean_text"] + " " + test_df ["emoji_text"]).str.strip()
```

```python
# ------------------------------
# 0. Setup
# ------------------------------
import torch
import evaluate
import re
from datasets import Dataset, DatasetDict
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from transformers import TrainingArguments, Trainer
from sklearn.model_selection import train_test_split

torch.cuda.empty_cache()

# ------------------------------
# 1. Label mapping (REQUIRED)
# ------------------------------
emotion_labels = ["anger", "disgust", "fear", "joy", "sadness", "surprise"]
label2id = {e: i for i, e in enumerate(emotion_labels)}
id2label = {i: e for e, i in label2id.items()}

# Add numeric labels to train_df
train_df["labels"] = train_df["emotion"].map(label2id)

# ------------------------------
# 2. Stratified split BEFORE converting to HF Dataset
# ------------------------------
train_df_split, valid_df_split = train_test_split(
    train_df,
    test_size=0.15,
    stratify=train_df["emotion"],
    random_state=42
)

# ------------------------------
# 3. Build HF Dataset dict
# ------------------------------
dataset = DatasetDict({
    "train": Dataset.from_pandas(train_df_split),
    "valid": Dataset.from_pandas(valid_df_split),
    "test":  Dataset.from_pandas(test_df)
})

# ------------------------------
# 4. Load tokenizer + model
# ------------------------------
MODEL_NAME = "roberta-base"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=6,
    label2id=label2id,
    id2label=id2label
)

# ------------------------------
# 5. Tokenize function
# ------------------------------
def tokenize(batch):
    return tokenizer(
        batch["full_text"],
        truncation=True,
        padding="max_length",
        max_length=128
    )

encoded = dataset.map(tokenize, batched=True)

# ------------------------------
# 6. Remove unused columns
# ------------------------------
cols_to_remove = [
    "text", "clean_text", "emoji_text", "emotion",
    "hashtags", "full_text", "__index_level_0__"
]

for split in ["train", "valid"]:
    encoded[split] = encoded[split].remove_columns(
        [c for c in cols_to_remove if c in encoded[split].column_names]
    )
    encoded[split].set_format(
        type="torch",
        columns=["input_ids", "attention_mask", "labels"]
    )

encoded["test"] = encoded["test"].remove_columns(
    [c for c in cols_to_remove if c in encoded["test"].column_names]
)
encoded["test"].set_format(
    type="torch",
    columns=["input_ids", "attention_mask"]
)

# ------------------------------
# 7. Final datasets ready for Trainer
# ------------------------------
ds_train = encoded["train"]
ds_valid = encoded["valid"]

# ------------------------------
# 8. Metrics
# ------------------------------
accuracy = evaluate.load("accuracy")
f1 = evaluate.load("f1")

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = logits.argmax(-1)
    return {
        "accuracy": accuracy.compute(predictions=preds, references=labels)["accuracy"],
        "f1_macro": f1.compute(predictions=preds, references=labels, average="macro")["f1"]
    }
```

```python
# ------------------------------
# 8. Quick TrainingArguments
# ------------------------------
training_args = TrainingArguments(
    output_dir="./roberta-quick",
    per_device_train_batch_size=80,     # ⬆ bigger batch
    per_device_eval_batch_size=80,
    gradient_accumulation_steps=1,      # simpler
    num_train_epochs=2,
    learning_rate=3e-5,
    weight_decay=0.01,
    logging_steps=100,
    save_total_limit=1,
    fp16=True,                          # ⬆ GPU speed
    gradient_checkpointing=True,        # ⬇ VRAM usage
)


# ------------------------------
# 9. Trainer
# ------------------------------
trainer = Trainer(
    model=model,
    args=training_args,
    tokenizer=tokenizer,
    train_dataset=ds_train,
    eval_dataset=ds_valid,
    compute_metrics=compute_metrics
)
```

```python
# ------------------------------
# 10. Train
# ------------------------------

trainer.train()

# ------------------------------
# 11. Evaluate
# ------------------------------
metrics = trainer.evaluate()
print("\n=== QUICK ROBERTA METRICS ===")
print(metrics)
```

```python
import re

def clean_for_transformer(text):
    if not isinstance(text, str):
        return ""

    # 1) Replace real newlines
    text = text.replace("\n", " ")

    # 2) Remove "escaped" \\n sequences
    text = re.sub(r"\\n+", " ", text)

    # 3) Replace usernames with placeholder
    #    @username → <user>
    text = re.sub(r"@\w+", "<user>", text)

    # 4) Remove URLs
    text = re.sub(r"http\S+|www\.\S+", "", text)

    # 5) Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
STRONG_CUES = [
    "disgust", "disgusting", "disgusted",
    "gross", "yuck", "ew", "eww", "ewww",
    "revolting", "repulsive",
    "nauseating", "nauseous", "vomit", "barf", "puke",
    "putrid", "rancid", "rotten",
    "foul", "stench", "stinky", "reeked",
    "filthy", "slimy", "moldy", "infested",
    "scum", "slut", "pedophilic",
    "abhorrent", "appalling", "repugnant", "detestable", "loathsome", "pedo", "hideous", "prick", "grossed"
]
```

```python
WEAK_CUES = list(set(DISGUST_CUES) - set(STRONG_CUES))
```

```python
import re

def flag_disgust(text):
    t = text.lower()

    # First check multi-word phrases (most reliable)
    multi_phrases = [c for c in DISGUST_CUES if " " in c]
    for phrase in multi_phrases:
        if phrase in t:
            return "[DISGUST] " + text

    # Then check strong cues using word boundaries
    for w in STRONG_CUES:
        if re.search(rf"\b{re.escape(w)}\b", t):
            return "[I'M DISGUSTED!] " + text

    # Then check weak cues — optional
    for w in WEAK_CUES:
        if re.search(rf"\b{re.escape(w)}\b", t):
            return "[DISGUST] " + text

    return text
```

```python
import re

def flag_sarcasm(text):
    if not isinstance(text, str):
        return text

    t = text.lower()

    # match common sarcasm cues
    sarcasm_patterns = [
        r"(?<!\w)\\s(?!\w)",        # literal "\s"
        r"(?<!\w)s\/(?!\w)",        # reddit sarcasm "s/"
        r"\/s(?!\w)",               # reddit "/s"
        r"(?<!\w)s\/$",             # trailing sarcasm
        r"#sarcasm",               # hashtag
        r"sarcastic"               # explicit word
    ]

    for p in sarcasm_patterns:
        if re.search(p, t):
            return "[THIS TEXT IS SARCASTIC] " + text

    return text
```

```python
# ============================================================
# SPECIALIST MODEL — Joy vs Anger (2-class RoBERTa)
# ============================================================

import torch
from datasets import Dataset, DatasetDict
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)
from sklearn.model_selection import train_test_split
import evaluate

torch.cuda.empty_cache()

print("\n=== Training Joy vs Anger Specialist Model ===\n")

# ------------------------------------------------------------
# 1. Filter ONLY joy + anger rows using the SAME split strategy
# ------------------------------------------------------------

df_j_vs_a = train_df[train_df["emotion"].isin(["joy", "anger"])].copy()

# Label mapping for ONLY 2 classes
jvsa_labels = ["anger", "joy"]
jvsa_label2id = {e: i for i, e in enumerate(jvsa_labels)}
jvsa_id2label = {i: e for e, i in jvsa_label2id.items()}

df_j_vs_a["labels"] = df_j_vs_a["emotion"].map(jvsa_label2id)

# Same style of split as your RoBERTa main model
jvsa_train_df, jvsa_valid_df = train_test_split(
    df_j_vs_a,
    test_size=0.15,
    stratify=df_j_vs_a["emotion"],
    random_state=42
)

# ------------------------------------------------------------
# 2. Build HuggingFace Dataset
# ------------------------------------------------------------
jvsa_dataset = DatasetDict({
    "train": Dataset.from_pandas(jvsa_train_df),
    "valid": Dataset.from_pandas(jvsa_valid_df)
})

# ------------------------------------------------------------
# 3. Load tokenizer + specialist model
# ------------------------------------------------------------
SPECIALIST_MODEL_NAME = "distilroberta-base"   # FAST + accurate

jvsa_tokenizer = AutoTokenizer.from_pretrained(SPECIALIST_MODEL_NAME)

jvsa_model = AutoModelForSequenceClassification.from_pretrained(
    SPECIALIST_MODEL_NAME,
    num_labels=2,
    label2id=jvsa_label2id,
    id2label=jvsa_id2label
)

# ------------------------------------------------------------
# 4. Tokenizer function
# ------------------------------------------------------------
def jvsa_tokenize(batch):
    texts = [t if isinstance(t, str) else "" for t in batch["full_text"]]
    return jvsa_tokenizer(
        texts,
        truncation=True,
        padding="max_length",
        max_length=99      # same as your main RoBERTa setup
    )

jvsa_encoded = jvsa_dataset.map(jvsa_tokenize, batched=True)

# ------------------------------------------------------------
# 5. Remove unused columns
# ------------------------------------------------------------
cols_remove = [
    "text", "clean_text", "emoji_text", "hashtags",
    "emotion", "__index_level_0__"
]

for split in ["train", "valid"]:
    jvsa_encoded[split] = jvsa_encoded[split].remove_columns(
        [c for c in cols_remove if c in jvsa_encoded[split].column_names]
    )
    jvsa_encoded[split].set_format(
        type="torch",
        columns=["input_ids", "attention_mask", "labels"]
    )

jvsa_train = jvsa_encoded["train"]
jvsa_valid = jvsa_encoded["valid"]

# ------------------------------------------------------------
# 6. Metrics
# ------------------------------------------------------------
jvsa_accuracy = evaluate.load("accuracy")
jvsa_f1 = evaluate.load("f1")

def jvsa_compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = logits.argmax(-1)
    return {
        "accuracy": jvsa_accuracy.compute(predictions=preds, references=labels)["accuracy"],
        "f1_macro": jvsa_f1.compute(predictions=preds, references=labels, average="macro")["f1"]
    }


```

```python
# ------------------------------------------------------------
# 7. TrainingArguments
# ------------------------------------------------------------
jvsa_args = TrainingArguments(
    output_dir="./joy-anger-specialist",
    per_device_train_batch_size=80,
    per_device_eval_batch_size=80,
    num_train_epochs=4,
    learning_rate=3e-5,
    weight_decay=0.01,
    fp16=True,
    logging_steps=50,
    save_total_limit=1
)

# ------------------------------------------------------------
# 8. Trainer
# ------------------------------------------------------------
jvsa_trainer = Trainer(
    model=jvsa_model,
    args=jvsa_args,
    tokenizer=jvsa_tokenizer,
    train_dataset=jvsa_train,
    eval_dataset=jvsa_valid,
    compute_metrics=jvsa_compute_metrics
)

# ------------------------------------------------------------
# 9. Train!
# ------------------------------------------------------------
jvsa_trainer.train()

# ------------------------------------------------------------
# 10. Evaluate
# ------------------------------------------------------------
jvsa_metrics = jvsa_trainer.evaluate()
print("\n=== JOY ↔ ANGER SPECIALIST METRICS ===")
print(jvsa_metrics)
```

```python
from torch.nn import CrossEntropyLoss

class WeightedTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False):
        labels = inputs.get("labels")

        # class weights: disgust is 4x more important
        weights = torch.tensor([1.0, 4.0, 1.0, 1.0, 1.0, 1.0]).to(labels.device)

        loss_fct = CrossEntropyLoss(weight=weights)
        outputs = model(**inputs)
        logits = outputs.get("logits")

        loss = loss_fct(logits, labels)
        return (loss, outputs) if return_outputs else loss
```

```python
# ------------------------------
# 8. TrainingArguments (with cosine LR + warmup)
# ------------------------------
training_args = TrainingArguments(
    output_dir="./roberta-quick",

    per_device_train_batch_size=80,
    per_device_eval_batch_size=80,
    gradient_accumulation_steps=1,
    num_train_epochs=10,

    # 🔥 OPTIMIZER + LR SCHEDULER IMPROVEMENTS
    learning_rate=3e-5,
    weight_decay=0.01,

    lr_scheduler_type="cosine",     # <--- NEW
    warmup_ratio=0.06,              # <--- NEW (6% of steps warmup)

    # LOGGING / SAVING
    logging_steps=100,
    save_total_limit=1,

    # PERFORMANCE
    fp16=True,
    gradient_checkpointing=True,
)

trainer = WeightedTrainer(
    model=model,
    args=training_args,
    tokenizer=tokenizer,
    train_dataset=ds_train,
    eval_dataset=ds_valid,
    compute_metrics=compute_metrics
)
```
